```python
#8) Write a program to create multiple methods with different return types
class Demo:

    def integer_method(self):
        return 10

    def string_method(self):
        return "Python"

    def float_method(self):
        return 5.5


d = Demo()

print(d.integer_method())
print(d.string_method())
print(d.float_method())
```

```python
#11) Write a program to cerate and call a parameter constructor
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name =", self.name)
        print("Age =", self.age)


s = Student("Jassmitha", 21)
s.display()
```

```python
#13) Write a program to call multiple constructors with a single constructor
class Demo:

    def __init__(self, a=None, b=None):

        if a is not None and b is not None:
            print("Two Arguments Constructor")
            print(a, b)

        elif a is not None:
            print("One Argument Constructor")
            print(a)

        else:
            print("No Argument Constructor")


d1 = Demo()
d2 = Demo(10)
d3 = Demo(10, 20)
```

```python
#14) Write a program to create and call different types of variables
class Demo:

    class_var = 100

    def __init__(self):
        self.instance_var = 200


d = Demo()

local_var = 300

print("Class Variable =", Demo.class_var)
print("Instance Variable =", d.instance_var)
print("Local Variable =", local_var)
```

```python
#15) Write a program to exolain the scope of variables
global_var = 50

class Demo:

    def show(self):

        local_var = 20

        print("Local Variable =", local_var)
        print("Global Variable =", global_var)


d = Demo()
d.show()
```

```python
#16) Program to show the use of Static and Non-static Methods
class Demo:

    @staticmethod
    def static_method():
        print("This is Static Method")

    def non_static_method(self):
        print("This is Non-Static Method")


Demo.static_method()

d = Demo()
d.non_static_method()
```
