```python
#11) Program to check whether a number is perfect number or not

num = int(input("Enter a number: "))

sum = 0

for i in range(1, num):

    if num % i == 0:
        sum = sum + i

if sum == num:
    print(num, "is a Perfect Number")

else:
    print(num, "is Not a Perfect Number")
```

```python
#13) Program to find biggest digit in a number
num = int(input("Enter a number: "))
big = 0
while num > 0:
    digit = num % 10
    if digit > big:
        big = digit
    num = num // 10
print("Biggest digit is:", big)
```

```python
#14) Program to find sum of digits using recursion
def sum_digits(n):
    if n == 0:
        return 0
    return (n % 10) + sum_digits(n // 10)
num = int(input("Enter a number: "))
result = sum_digits(num)
print("Sum of digits =", result)
```

```python
#19) Program to add even digits and odd digits separately
num = input("Enter a number: ")
even_sum = 0
odd_sum = 0
for i in num:
    digit = int(i)
    if digit % 2 == 0:
        even_sum = even_sum + digit
    else:
        odd_sum = odd_sum + digit
print("Sum of even digits =", even_sum)
print("Sum of odd digits =", odd_sum)
```

```python
#20) Program to reverse a number
num = int(input("Enter a number: "))
reverse = 0
while num > 0:
    digit = num % 10
    reverse = (reverse * 10) + digit
    num = num // 10
print("Reversed number =", reverse)
```

```python
#21) Program to print prime numbers between m and n
m = int(input("Enter starting number: "))
n = int(input("Enter ending number: "))
print("Prime numbers are:")
for num in range(m, n + 1):
    if num > 1:
        prime = True
        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
        if prime:
            print(num, end=" ")
```

```python
#22) Program to print first n prime numbers
n = int(input("Enter value of n: "))
count = 0
num = 2
print("Prime numbers are:")
while count < n:
    prime = True
    for i in range(2, num):
        if num % i == 0:
            prime = False
            break
    if prime:
        print(num, end=" ")
        count = count + 1
    num = num + 1
```

```python
#23) Program to print perfect numbers between m and n
m = int(input("Enter starting number: "))
n = int(input("Enter ending number: "))
print("Perfect numbers are:")
for num in range(m, n + 1):
    sum = 0
    for i in range(1, num):
        if num % i == 0:
            sum = sum + i
    if sum == num:
        print(num, end=" ")
```

```python
#25) Program to print Armstrong numbers from 1 to 1000
for num in range(1, 1001):
    temp = num
    sum = 0
    digits = len(str(num))
    while temp > 0:
        digit = temp % 10
        sum = sum + (digit ** digits)
        temp = temp // 10
    if sum == num:
        print(num, end=" ")
```
