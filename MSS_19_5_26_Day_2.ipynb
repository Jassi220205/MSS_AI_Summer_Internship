```python
#25) Program to print Armstrong numbers from 1 to 1000
for num in range(1, 1001):
    temp = num
    sum = 0
    digits = len(str(num))
    while temp > 0:
        digit = temp % 10
        sum = sum + (digit ** digits)
        temp = temp // 10
    if sum == num:
        print(num, end=" ")
```

```python
#32) Program to sort array elements in descending order
arr = []
n = int(input("Enter number of elements: "))
print("Enter array elements:")
for i in range(n):
    value = int(input())
    arr.append(value)
for i in range(n):
    for j in range(i + 1, n):
        if arr[i] < arr[j]:
            temp = arr[i]
            arr[i] = arr[j]
            arr[j] = temp
print("Array in descending order:")
for i in arr:
    print(i, end=" ")
```

```python
#44) Program to demonstrate public, protected and private access modifiers
class Student:
    # Public variable
    name = "Jas"
    # Protected variable
    _course = "Python"
    # Private variable
    __marks = 95
    # Method to display values
    def display(self):
        print("Name =", self.name)
        print("Course =", self._course)
        print("Marks =", self.__marks)
# Creating object
obj = Student()
# Accessing public variable
print("Public Variable:")
print(obj.name)
# Accessing protected variable
print("\nProtected Variable:")
print(obj._course)
# Accessing private variable using method
print("\nPrivate Variable:")
obj.display()
```
