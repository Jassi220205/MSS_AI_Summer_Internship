```python
#1. Write a Python program to find all three, four, and five-character words in a string.
import re

text = "Python is fun and regex helps coding"

result = re.findall(r'\b\w{3,5}\b', text)

print(result)
```

```python
#2. Write a Python script to transform a camelCase identifier into snake_case and then print the result.
import re

text = "myVariableName"

result = re.sub(r'([A-Z])', r'_\1', text).lower()

print(result)
```

```python
#3. Write a Python program to find URLs in a string.
import re

text = "Visit https://google.com and http://github.com"

urls = re.findall(r'https?://\S+', text)

print(urls)
```

```python
#4. Write a Python program to remove everything except alphanumeric characters from a string.
import re

text = "Hello@123#Python!"

result = re.sub(r'[^a-zA-Z0-9]', '', text)

print(result)
```

```python
#5. Write a Python program to split a string with multiple delimiters.
import re

text = "Apple,Orange;Banana:Grapes"

result = re.split(r'[,;:]', text)

print(result)
```

```python
#6. Write a Python program to replace whitespaces with an underscore and vice versa.
import re

text = "Hello Python World"

result = re.sub(r'\s', '_', text)

print(result)
```

```python
#7. Write a Python program that matches a string that has an a followed by zero or more b's.
import re

text = "abbb"

if re.fullmatch(r'ab*', text):
    print("Matched")
```

```python
#8. Write a Python program that matches a string that has an a followed by zero or one 'b'.
import re

text = "ab"

if re.fullmatch(r'ab?', text):
    print("Matched")
```

```python
#9. Write a Python program that matches a string with an a followed by two to three 'b'.
import re

text = "abbb"

if re.fullmatch(r'ab{2,3}', text):
    print("Matched")
```

```python
#10. Write a Python program that matches a string with an a ' followed by anything ending in 'b'.
import re

text = "a123xyzb"

if re.fullmatch(r'a.*b', text):
    print("Matched")
```

```python
#11. Write a Python program that matches a word containing 'z', not the start or end of the word.
import re

text = "amazing frozen zebra"

result = re.findall(r'\b\w+z\w+\b', text)

print(result)
```

```python
#12. Write a Python program that starts each string with a specific number.
import re

text = "123Python"

if re.match(r'^123', text):
    print("Starts with 123")
```

```python
#13. Write a Python program to get the year, month, and date from a URL.
import re

url = "https://example.com/2025/07/15/article"

result = re.search(r'(\d{4})/(\d{2})/(\d{2})', url)

print("Year:", result.group(1))
print("Month:", result.group(2))
print("Date:", result.group(3))
```

```python
#14. Write a Python program to separate and print the numbers and their position in a given string.
import re

text = "abc123xyz45"

for match in re.finditer(r'\d+', text):
    print("Number:", match.group())
    print("Position:", match.start())
```

```python
#15. Write a Python program to replace with given number of occurrences of space, comma, or dot with a colon.
import re

text = "Python, Java. AI ML"

result = re.sub(r'[ ,.]+', ':', text)

print(result)
```
