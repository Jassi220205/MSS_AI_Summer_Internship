```python
#1. Write a Python function that accepts a string and counts the number of upper and lowercase letters
def count_case(text):
  upper=0
  lower=0
  for ch in text:
    if ch.isupper():
      upper+=1
    elif ch.islower():
      lower+=1
  print("No.of upper case characters:",upper)
  print("No. of lower case characters:",lower)
count_case("ThequickBrownFox")
```

```python
#2. Write a Python function that takes a list and returns a new list with distinct elements from the first list
def unique_list(lst):
  unique=[]
  for item in lst:
    if item not in unique:
      unique.append(item)
  return unique
print(unique_list([1,2,3,3,3,3,4,5]))
```

```python
#3. Write a Python function that checks whether a passed string is a palindrome or not.
def is_palindrome(s):
  cleaned_s = "".join(char.lower() for char in s if char.isalnum())
  return cleaned_s == cleaned_s[::-1]

# Test cases
print(f"'madam' is a palindrome: {is_palindrome('madam')}")
print(f"'nurses run' is a palindrome: {is_palindrome('nurses run')}")
print(f"'hello' is a palindrome: {is_palindrome('hello')}")
print(f"'A man, a plan, a canal: Panama' is a palindrome: {is_palindrome('A man, a plan, a canal: Panama')}")
```

```python
#4. Write a Python program to square the elements of a list using the map() function.

def square(number):
  return number ** 2

my_list = [1, 2, 3, 4, 5]
squared_list = list(map(square, my_list))

print(f"Original List: {my_list}")
print(f"Squared List: {squared_list}")
```

```python
#5. Write a Python program to add two given lists and find the difference between them.
list1 = [6, 5, 3, 9]
list2 = [0, 1, 7, 7]

def add_and_diff(x, y):
  return (x + y, x - y)

result = list(map(add_and_diff, list1, list2))
print(f"Result: {result}")
```

```python
#8. Write a Python function to check whether a number is "Perfect" or not. In number theory, a perfect number is a positive integer that is equal to the sum of its proper positive divisors, that is, the sum of its positive divisors excluding the number itself (also known as its aliquot sum). Equivalently, a perfect number is a number that is half the sum of all of its positive divisors (including itself)
def is_perfect_number(num):
    if num <= 0:
        return False
    sum_of_divisors = 0
    for i in range(1, num):
        if num % i == 0:
            sum_of_divisors += i
    return sum_of_divisors == num
print(is_perfect_number(496))
```

```python
#11. Write a Python program that creates a list of names and uses the filter function to extract names that start with a vowel (A, E, I, O, U).
names = ['Alice', 'Bob', 'Eve', 'David', 'Ursula', 'Owen', 'Frank', 'Anna']
vowels = 'AEIOU'
vowel_names = list(filter(lambda name: name[0].upper() in vowels, names))
print(f"Original list of names: {names}")
print(f"Names starting with a vowel: {vowel_names}")
```
