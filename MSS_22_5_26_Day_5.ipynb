```python
#1. Create a generator function that yields Fibonacci numbers up to a specified limit.
def fibonacci(limit):

    a = 0
    b = 1

    while a <= limit:
        yield a

        temp = a + b
        a = b
        b = temp


limit = int(input("Enter limit: "))

for num in fibonacci(limit):
    print(num)
```

```python
#3. Write a generator that produces an infinite sequence of prime numbers.
def is_prime(num):

    if num < 2:
        return False

    for i in range(2, num):

        if num % i == 0:
            return False

    return True


def prime_generator():

    num = 2

    while True:

        if is_prime(num):
            yield num

        num += 1


gen = prime_generator()

for i in range(10):
    print(next(gen))
```

```python
#4. Create a generator function that flattens nested lists of arbitrary depth.
def flatten(data):

    for item in data:

        if type(item) == list:

            for x in flatten(item):
                yield x

        else:
            yield item


nested = [1, [2, 3], [4, [5, 6]], 7]

for value in flatten(nested):
    print(value)
```

```python
#5. Create a decorator to measure and print the execution time of functions.
import time

def timer(func):

    def wrapper():

        start = time.time()

        func()

        end = time.time()

        print("Execution Time:", end - start)

    return wrapper


@timer
def display():

    for i in range(1000000):
        pass


display()
```

```python
#6. Create a retry decorator that attempts to execute a function multiple times if it fails.
def retry(times):

    def decorator(func):

        def wrapper():

            for i in range(times):

                try:
                    return func()

                except Exception:
                    print("Retrying...")

            print("Function failed")

        return wrapper

    return decorator


count = 0

@retry(3)
def test():

    global count

    count += 1

    if count < 3:
        raise Exception

    print("Function executed")


test()
```

```python
#7. Build a decorator to validate function arguments against specified criteria.
def validate(func):

    def wrapper(age):

        if age < 0:
            print("Invalid age")

        else:
            func(age)

    return wrapper


@validate
def display(age):

    print("Age is:", age)


display(20)
display(-5)
```

```python
#8. Create a class-based decorator that counts how many times a function is called.
class Counter:

    def __init__(self, func):

        self.func = func
        self.count = 0

    def __call__(self):

        self.count += 1

        print("Called", self.count, "times")

        self.func()


@Counter
def hello():

    print("Hello")


hello()
hello()
hello()
```

```python
#9. Implement a parameterized decorator that logs function calls with a customizable message format.
def logger(message):

    def decorator(func):

        def wrapper():

            print(message)

            func()

        return wrapper

    return decorator


@logger("Function is executing")
def greet():

    print("Welcome")


greet()
```

```python
#10. Create a decorator that makes a function run asynchronously.
import threading

def async_function(func):

    def wrapper():

        thread = threading.Thread(target=func)

        thread.start()

    return wrapper


@async_function
def task():

    for i in range(5):
        print("Running task")


task()

print("Main program continues")
```

```python
#11(b). Create a Custom Iterator
class Numbers:

    def __init__(self, limit):

        self.limit = limit
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):

        if self.num <= self.limit:

            value = self.num
            self.num += 1

            return value

        else:
            raise StopIteration


obj = Numbers(5)

for i in obj:
    print(i)
```

```python
#11(c). Create a Infinite Iterator
class Infinite:

    def __init__(self):

        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):

        value = self.num
        self.num += 1

        return value


obj = Infinite()

for i in obj:

    if i > 10:
        break

    print(i)
```
