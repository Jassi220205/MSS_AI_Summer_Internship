```python
#2. Write a program to handle exceptions with Multiple Except Block.
try:
    num = int(input("Enter a number: "))
    result = 10 / num

    print("Result:", result)

except ValueError:
    print("Error: Invalid input. Please enter numbers only")

except ZeroDivisionError:
    print("Error: Cannot divide by zero")
```

```python
#3. Write a program to handle exceptions with  Nested Try.
try:
    print("Outer try block")

    try:
        num = int(input("Enter a number: "))
        result = 10 / num
        print("Result:", result)

    except (ZeroDivisionError, ValueError) as e:
        print("Inner Exception's error invoked", e)

except ValueError:
    print("Outer except block")
```

```python
#4. Write a program to execute a statement using Finally Block
try:
    print("Inside try block")
    x = 10 / 2
    print(x)

finally:
    print("Finally block executed")
```

```python
#5. Write a program to execute a statement using raise Block.
try:
    age = int(input("Enter your age: "))

    if age < 18:
        raise ValueError("Age must be 18 or above")

    print("Access granted")

except ValueError as e:
    print("Error:", e)
```

```python
#create a custom exception to check account number and password if valid display options-withdraw(atleast 500), deposit(atleast 1000), balance, exit and if invalid- raise exceptions, give 3 password tries, if all 3 fail, exit
class InvalidAccountNumber(Exception):
    pass


class DepositError(Exception):
    pass


class WithdrawError(Exception):
    pass


class BankAccount:
    def __init__(self, acc_no, password, balance=0):
        self.acc_no = acc_no
        self.password = password
        self.balance = balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Enter Valid Amount")

        elif amount < 500:
            raise WithdrawError("Minimum withdrawal amount is 500")

        elif amount > self.balance:
            raise ValueError("Insufficient Balance")

        else:
            self.balance -= amount
            return self.balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Enter Valid Amount")

        elif amount < 1000:
            raise DepositError("Minimum deposit amount is 1000")

        else:
            self.balance += amount
            return self.balance

    def check_balance(self):
        return self.balance

    @staticmethod
    def validate_account(acc_no):
        if not acc_no.isdigit():
            raise InvalidAccountNumber(
                "Account number must contain only digits"
            )

        if not (8 <= len(acc_no) <= 18):
            raise InvalidAccountNumber(
                "Account number must be 8-18 digits long"
            )

        return True


def main():

    sample_accns = {
        "12345678": "pass123",
        "9876543210": "bank456"
    }

    balances = {
        "12345678": 5000,
        "9876543210": 10000
    }

    try:
        acc_no = input("Enter Account Number: ")

        BankAccount.validate_account(acc_no)

        if acc_no not in sample_accns:
            raise InvalidAccountNumber("Account not found")

        account = BankAccount(
            acc_no,
            sample_accns[acc_no],
            balances[acc_no]
        )

        for attempt in range(3):

            pwd = input(
                f"Enter Password. Attempt {attempt + 1}/3: "
            )

            if pwd == account.password:
                print("Login Successful!")
                break

            else:
                remaining = 2 - attempt

                if remaining > 0:
                    print(
                        f"Wrong Password. {remaining} tries left"
                    )

                else:
                    raise InvalidAccountNumber(
                        "3 failed password attempts. Exiting"
                    )

    except InvalidAccountNumber as e:
        print(f"Error: {e}")
        return

    except ValueError as e:
        print(f"Error: {e}")
        return

    while True:

        print("\n--- MENU ---")
        print("1. Withdraw")
        print("2. Deposit")
        print("3. Balance")
        print("4. Exit")

        choice = input("Enter choice: ")

        try:

            if choice == '1':

                amt = float(
                    input("Enter amount to withdraw: ")
                )

                new_bal = account.withdraw(amt)

                print(
                    f"Withdrawn Successfully. "
                    f"New balance: {new_bal}"
                )

            elif choice == '2':

                amt = float(
                    input("Enter amount to deposit: ")
                )

                new_bal = account.deposit(amt)

                print(
                    f"Deposited Successfully. "
                    f"New balance: {new_bal}"
                )

            elif choice == '3':

                print(
                    f"Current balance: "
                    f"{account.check_balance()}"
                )

            elif choice == '4':

                print("Thank you. Exiting...")
                break

            else:
                print("Invalid choice")

        except ValueError as e:
            print(f"Error: {e}")

        except DepositError as e:
            print(f"Deposit Error: {e}")

        except WithdrawError as e:
            print(f"Withdraw Error: {e}")


if __name__ == "__main__":
    main()
```
