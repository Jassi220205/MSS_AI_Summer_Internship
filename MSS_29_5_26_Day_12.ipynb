```python
#18. Program to count vowels, consonants, digits, and special characters present in a text file separately.
file = open("data.txt", "r")

content = file.read()

vowels = consonants = digits = special = 0

for ch in content:
    if ch.lower() in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Special characters:", special)

file.close()
```

```python
#25. Program to identify and display the most repeated word present in a text file with its frequency.
file = open("data.txt", "r")

words = file.read().split()

frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

max_word = max(frequency, key=frequency.get)

print("Most repeated word:", max_word)
print("Frequency:", frequency[max_word])

file.close()
```

```python
#28. Program to identify repeated lines in a file and display the duplicated lines only once.
file = open("data.txt", "r")

lines = file.readlines()

repeated = set()

for line in lines:
    if lines.count(line) > 1:
        repeated.add(line.strip())

print("Repeated lines:")
for line in repeated:
    print(line)

file.close()
```

```python
#30. Program to find and display all possible anagram words available in a text file.
file = open("data.txt", "r")

words = file.read().split()

anagrams = {}

for word in words:
    key = ''.join(sorted(word))

    if key not in anagrams:
        anagrams[key] = []

    anagrams[key].append(word)

for group in anagrams.values():
    if len(group) > 1:
        print(group)

file.close()
```

```python
#33. Program to identify and display the top 10 most frequently used words present in a text file.
file = open("data.txt", "r")

words = file.read().split()

frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

sorted_words = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

print("Top 10 words:")

for word, count in sorted_words[:10]:
    print(word, ":", count)

file.close()
```
